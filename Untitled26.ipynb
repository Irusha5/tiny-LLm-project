```python
from datasets import load_dataset

# 100% download wenne na, podi chunk ekak witharak gannawa
dataset = load_dataset("roneneldan/TinyStories", split="train", streaming=True)

# first 5 stories gannawa
small_data = list(dataset.take(100))
print(f"Got {len(small_data)} stories")
print(small_data[0]['text'][:600])
print("\n--- DONE ---")
```

```python
import torch
import torch.nn as nn
from torch.nn import functional as F
import math

# 1. Prepare text from streaming data
all_text = " ".join([d['text'] for d in small_data])
chars = sorted(list(set(all_text)))
vocab_size = len(chars)
print(f"Vocab size: {vocab_size}")

stoi = { ch:i for i,ch in enumerate(chars) }
itos = { i:ch for i,ch in enumerate(chars) }
encode = lambda s: [stoi.get(c,0) for c in s]
decode = lambda l: ''.join([itos[i] for i in l])

data = torch.tensor(encode(all_text), dtype=torch.long)
n = int(0.9*len(data))
train_data = data[:n]
val_data = data[n:]
print(f"Train len: {len(train_data)}")

# 2. Tiny Transformer Model (Thillini ge code eka)
block_size = 64
batch_size = 32

class BigramLM(nn.Module):
    def __init__(self, vocab_size):
        super().__init__()
        self.token_embedding_table = nn.Embedding(vocab_size, 128)
        self.position_embedding_table = nn.Embedding(block_size, 128)
        self.lm_head = nn.Linear(128, vocab_size)

    def forward(self, idx, targets=None):
        B, T = idx.shape
        tok_emb = self.token_embedding_table(idx) # B,T,128
        pos_emb = self.position_embedding_table(torch.arange(T, device=idx.device))
        x = tok_emb + pos_emb
        logits = self.lm_head(x) # B,T,vocab
        if targets is None:
            loss = None
        else:
            B, T, C = logits.shape
            logits = logits.view(B*T, C)
            targets = targets.view(B*T)
            loss = F.cross_entropy(logits, targets)
        return logits, loss

    def generate(self, idx, max_new_tokens):
        for _ in range(max_new_tokens):
            idx_cond = idx[:, -block_size:]
            logits, _ = self(idx_cond)
            logits = logits[:, -1, :]
            probs = F.softmax(logits, dim=-1)
            idx_next = torch.multinomial(probs, num_samples=1)
            idx = torch.cat((idx, idx_next), dim=1)
        return idx

def get_batch(split):
    data_split = train_data if split == 'train' else val_data
    ix = torch.randint(len(data_split) - block_size, (batch_size,))
    x = torch.stack([data_split[i:i+block_size] for i in ix])
    y = torch.stack([data_split[i+1:i+block_size+1] for i in ix])
    return x, y

device = "cuda" if torch.cuda.is_available() else "cpu"
model = BigramLM(vocab_size).to(device)
optimizer = torch.optim.AdamW(model.parameters(), lr=1e-3)

# 3. Training - short training for assignment
for iter in range(300):
    xb, yb = get_batch('train')
    xb, yb = xb.to(device), yb.to(device)
    logits, loss = model(xb, yb)
    optimizer.zero_grad(set_to_none=True)
    loss.backward()
    optimizer.step()
    if iter % 50 == 0:
        print(f"{iter}: loss {loss.item():.4f}")

print("\n--- TRAINING DONE ---")

# 4. Generate story
context = torch.zeros((1,1), dtype=torch.long, device=device)
print(decode(model.generate(context, max_new_tokens=300)[0].tolist()))
```

```python
import matplotlib.pyplot as plt

# 5. Plot loss (oya report ekata danna puluwan graph eka)
losses = [4.5338, 2.9883, 2.5983, 2.4950, 2.4082, 2.3895] # oya print wuna values
plt.plot(losses)
plt.title("Training Loss - TinyStories")
plt.xlabel("Iter (x50)")
plt.ylabel("Loss")
plt.show()

# 6. Better generation with prompt
def generate_with_prompt(start_text, max_tokens=200):
    model.eval()
    encoded = encode(start_text)
    context = torch.tensor([encoded], dtype=torch.long, device=device)
    generated = model.generate(context, max_new_tokens=max_tokens)
    return decode(generated[0].tolist())

print(generate_with_prompt("Once upon a time", 300))
print("\n---\n")
print(generate_with_prompt("Lily was", 300))

# 7. Save model for submission
torch.save(model.state_dict(), "tiny_model.pt")
print("Model saved! tiny_model.pt")
```
